```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return (x + 3)**2

def df(x):
    return 2 * (x + 3)

cur_x = 2.0
learning_rate = 0.1
precision = 1e-6
max_iters = 1000

previous_step_size = 1.0
iters = 0
x_history = [cur_x]
y_history = [f(cur_x)]

print(f"Initial Starting Point: x = {cur_x}, y = {f(cur_x)}")
print("-" * 55)
print(f"{'Iteration':<10} | {'x Value':<15} | {'f(x) Value':<15} | {'Gradient':<10}")
print("-" * 55)

while previous_step_size > precision and iters < max_iters:
    prev_x = cur_x
    grad = df(prev_x)

    cur_x = prev_x - learning_rate * grad

    previous_step_size = abs(cur_x - prev_x)
    iters += 1

    x_history.append(cur_x)
    y_history.append(f(cur_x))

    if iters <= 10 or iters % 5 == 0:
        print(f"{iters:<10} | {cur_x:<15.6f} | {f(cur_x):<15.6f} | {grad:<10.6f}")

print("-" * 55)
print("\n" + "="*50)
print("              RESULTS SUMMARY              ")
print("="*50)
print(f"Total Iterations    : {iters}")
print(f"Local Minima at x   : {cur_x:.6f}")
print(f"Minimum Value y     : {f(cur_x):.6f}")
print("="*50)

x_vals = np.linspace(-9, 3, 400)
y_vals = f(x_vals)

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, 'b-', label='y = (x + 3)^2', linewidth=2)
plt.plot(x_history, y_history, 'ro--', label='Gradient Descent Trajectory', markersize=6)
plt.scatter(cur_x, f(cur_x), color='green', s=150, zorder=5, label=f'Local Minima (~{cur_x:.2f}, {f(cur_x):.2f})')

plt.title("Gradient Descent Optimization on y = (x + 3)^2", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)
plt.show()
```
