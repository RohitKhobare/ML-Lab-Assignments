```python
import pandas as pd
import numpy as np
import os
import kagglehub

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import (confusion_matrix, accuracy_score, precision_score,
                             recall_score, f1_score, mean_squared_error,
                             mean_absolute_error, r2_score)

# ==========================================
# 1. DOWNLOAD DATASET VIA KAGGLEHUB (FOR COLAB)
# ==========================================
print("Downloading Email Spam Dataset from Kaggle...")
path = kagglehub.dataset_download("balaka18/email-spam-classification-dataset-csv")

# Locate and read the downloaded CSV file
csv_file = [os.path.join(path, f) for f in os.listdir(path) if f.endswith('.csv')][0]
df = pd.read_csv(csv_file)
print(f"Dataset loaded successfully! Initial shape: {df.shape}")

# ==========================================
# 2. PRE-PROCESSING & FEATURE SELECTION
# ==========================================
# Drop non-predictive identifier column if present
if 'Email No.' in df.columns:
    df = df.drop(columns=['Email No.'])

X = df.drop(columns=['Prediction'])
y = df['Prediction']

# Stratified Train-Test Split (80:20 ratio)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Feature Scaling (Crucial for KNN and SVM distance calculations)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ==========================================
# 3. MODEL TRAINING & PREDICTION
# ==========================================
print("\nTraining K-Nearest Neighbors Classifier...")
knn = KNeighborsClassifier(n_neighbors=5, metric='minkowski', p=2)
knn.fit(X_train_scaled, y_train)
y_pred_knn = knn.predict(X_test_scaled)

print("Training Support Vector Machine Classifier...")
svm = SVC(kernel='rbf', C=1.0, random_state=42)
svm.fit(X_train_scaled, y_train)
y_pred_svm = svm.predict(X_test_scaled)

# ==========================================
# 4. EVALUATION METRICS COMPARISON
# ==========================================
def evaluate_classifier(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred)
    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred)
    rec = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)

    # Continuous / Statistical error metrics
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)

    return cm, acc, prec, rec, f1, mse, rmse, mae, r2

cm_knn, acc_k, prec_k, rec_k, f1_k, mse_k, rmse_k, mae_k, r2_k = evaluate_classifier(y_test, y_pred_knn)
cm_svm, acc_s, prec_s, rec_s, f1_s, mse_s, rmse_s, mae_s, r2_s = evaluate_classifier(y_test, y_pred_svm)

# Output Results Table
print("\n" + "="*70)
print(f"{'Metric':<18} | {'K-Nearest Neighbors':<22} | {'Support Vector Machine':<22}")
print("="*70)
print(f"{'Accuracy':<18} | {acc_k:<22.4f} | {acc_s:<22.4f}")
print(f"{'Precision':<18} | {prec_k:<22.4f} | {prec_s:<22.4f}")
print(f"{'Recall':<18} | {rec_k:<22.4f} | {rec_s:<22.4f}")
print(f"{'F1 Score':<18} | {f1_k:<22.4f} | {f1_s:<22.4f}")
print(f"{'MSE':<18} | {mse_k:<22.4f} | {mse_s:<22.4f}")
print(f"{'RMSE':<18} | {rmse_k:<22.4f} | {rmse_s:<22.4f}")
print(f"{'MAE':<18} | {mae_k:<22.4f} | {mae_s:<22.4f}")
print(f"{'R2 Score':<18} | {r2_k:<22.4f} | {r2_s:<22.4f}")
print("="*70)

print("\nKNN Confusion Matrix:\n", cm_knn)
print("\nSVM Confusion Matrix:\n", cm_svm)
```
